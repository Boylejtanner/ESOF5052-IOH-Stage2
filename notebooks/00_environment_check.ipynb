# Environment verification

Checks the selected Python environment and essential packages without loading clinical data.

Run the setup cell first. Full processing and training require locally provisioned datasets; the quick start uses synthetic tests and saved aggregate results.


```python
# Load the libraries and project settings needed for this step.
import os
import sys
from importlib.metadata import PackageNotFoundError, version

print(f"Python: {sys.version.split()[0]}")
print(f"Interpreter: {sys.executable}")
for package in ("numpy", "pandas", "pyarrow", "scipy", "scikit-learn", "vitaldb", "torch", "pyspark"):
    try:
        print(f"{package}: {version(package)}")
    except PackageNotFoundError:
        print(f"{package}: not installed; see environment/ dependency groups")
print(f"JAVA_HOME: {os.getenv('JAVA_HOME', 'not configured; needed only for Spark')}")
```
